# Ejercicio 3: Operador de cruzamiento de dos puntos

Se comparan el cruzamiento estándar de un punto y el cruzamiento de dos puntos. En el segundo operador se seleccionan dos índices aleatorios y se intercambia únicamente el segmento central entre los padres.

```python
import random

def cruzamiento_un_punto(padre_1, padre_2, rng=None):
    """Intercambia el sufijo de los padres después de un corte."""
    rng = rng or random
    if len(padre_1) != len(padre_2) or len(padre_1) < 2:
        raise ValueError('Los padres deben tener la misma longitud y al menos dos genes.')
    corte = rng.randint(1, len(padre_1) - 1)
    hijo_1 = padre_1[:corte] + padre_2[corte:]
    hijo_2 = padre_2[:corte] + padre_1[corte:]
    return hijo_1, hijo_2

def cruzamiento_dos_puntos(padre_1, padre_2, rng=None):
    """Intercambia el segmento entre dos cortes aleatorios."""
    rng = rng or random
    if len(padre_1) != len(padre_2) or len(padre_1) < 3:
        raise ValueError('Los padres deben tener la misma longitud y al menos tres genes.')

    corte_1, corte_2 = sorted(rng.sample(range(1, len(padre_1)), 2))
    hijo_1 = (padre_1[:corte_1] + padre_2[corte_1:corte_2] + padre_1[corte_2:])
    hijo_2 = (padre_2[:corte_1] + padre_1[corte_1:corte_2] + padre_2[corte_2:])
    return hijo_1, hijo_2, (corte_1, corte_2)
```

```python
class CortesDeterministas:
    """Permite probar el operador sin depender de la aleatoriedad."""
    def __init__(self, cortes):
        self.cortes = iter(cortes)

    def sample(self, poblacion, cantidad):
        return [next(self.cortes) for _ in range(cantidad)]

padre_1 = [1, 1, 1, 1, 1, 1]
padre_2 = [0, 0, 0, 0, 0, 0]
hijo_1, hijo_2, cortes = cruzamiento_dos_puntos(
    padre_1, padre_2, CortesDeterministas([2, 5])
)

print(f'Cortes seleccionados: {cortes}')
print(f'Hijo 1: {hijo_1}')
print(f'Hijo 2: {hijo_2}')

assert cortes == (2, 5)
assert hijo_1 == [1, 1, 0, 0, 0, 1]
assert hijo_2 == [0, 0, 1, 1, 1, 0]
```

## Resultado

Con dos puntos de corte se conserva el prefijo y el sufijo de cada padre, mientras que solo se intercambia el segmento central. Esto permite combinar bloques de genes sin reemplazar todo el resto del cromosoma.
